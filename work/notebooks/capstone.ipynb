## 1. Question

## Predicting SEO Content Decay: Moving from Static Rules to Machine Learning

**Abstract**
Content decay causes significant organic traffic loss if not identified and refreshed early. We utilized 30,000 anonymized historical performance records to predict whether a page's organic traffic trend is actively decaying. By training a Random Forest Classifier and validating it with a rigorous, grouped-client split, the model achieved a 0.9529 ROC-AUC. This represents a measured +0.27 AUC improvement over legacy heuristic baselines. This model serves as a directional decision-support tool, allowing SEO teams to prioritize editorial resources for content refreshes before major traffic bleed occurs.

**Introduction & Problem Statement**
Content strategy teams often waste hours manually auditing thousands of URLs, or worse, they rely on reactive reporting and only notice a problem *after* revenue drops. A predictive model shifts this workflow from reactive to proactive. By accurately scoring the decay risk of historical content, teams can systematically allocate their time to the pages where a refresh will provide the most measured value.


### 2. Data

*   **Source & Size:** The model was trained on 30,000 rows of production search data.
*   **Time Window:** Features represent 90-day and 30-day historical aggregations prior to the target measurement window.
*   **Exclusions:** Features like `trend_pct` and `trend_direction` were explicitly excluded from the feature set to prevent data leakage, as they directly compute the target variable. Identifier columns (`content_id`, `client_id`) were excluded from training to prevent the model from memorizing specific URLs.
*   **Public-Safety:** The dataset is fully anonymized. No client names, private URLs, or proprietary queries are exposed.


### 3. Methodology

*   **Label Definition:** The target variable (`needs_refresh`) was defined as a binary classification: `1` if the page's subsequent traffic trend (`trend_pct`) was negative, and `0` if it was flat or positive.
*   **Baseline:** The model was evaluated against a legacy heuristic rule that ranked items based purely on high visibility combined with low engagement: `impressions_90d * (1 - ctr)`.
*   **Model:** Random Forest Classifier.
*   **Validation Design (Honest Split):** To prove the model generalizes to unseen data, validation was performed using `GroupKFold` grouped by `client_id`. This ensures a client's data is either entirely in the training set or entirely in the validation set, preventing the model from simply memorizing client-specific behaviors.
*   **Leakage Checks:** Pearson correlation checks confirmed no numeric features had a suspiciously high correlation (>0.85) with the target, verifying the model relies on historical signals, not future data.


### 4. Results

**Observation:** The ML model provides a substantial, measured improvement in identifying underperforming content compared to the static rule-based approach, successfully capturing non-linear relationships between historical visibility, click-through rates, and future traffic decay.

### 5. Limitations & Honest Framing

*   **No Causal Guarantees:** This model predicts the *risk* of decay; it does not prove causation. Flagging a page for a refresh does not guarantee that changing a meta title will magically restore traffic.
*   **External Volatility:** The model does not understand external market context, sudden viral trends, or massive Google Core Algorithm updates.
*   **Cold Start Problem:** Pages with less than 90 days of historical data cannot be accurately scored by this model.

### 6. Ranked Recommendations (Action Playbook)

The model output is translated into a Content Action Playbook queue for the editorial team:

1.  **HIGH_VISIBILITY_DECAY:** (High Risk >75% + High Impressions). *Action:* Priority manual audit. This prevents the largest total traffic losses.
2.  **POOR_ENGAGEMENT_LEAK:** (High Risk >75% + Low CTR). *Action:* Standard refresh focusing strictly on meta descriptions and search-intent alignment.
3.  **STANDARD_DECAY:** (Moderate Risk 60-75%). *Action:* Batch review for lightweight, scheduled updates.

*Note: Legal pages, privacy policies, and core brand homepages are strictly excluded from automated refresh queues.*
